```python
# Setup the Jupyter version of Dash
# from jupyter_dash import JupyterDash
from dash import Dash

# Configure the necessary Python module imports for dashboard components
import dash_leaflet as dl
from dash import dcc, html
import plotly.express as px
from dash import dash_table
from dash.dependencies import Input, Output, State
import base64
# JupyterDash.infer_jupyter_proxy_config()

# Configure OS routines
import os

# Configure the plotting routines
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# Import CRUD Python module
from CRUD_Python_Module import AnimalShelter

###########################
# Data Manipulation / Model
###########################

username = "aacuser"
password = "password123"

# Connect to database via CRUD Module
db = AnimalShelter(username, password)

# class read method must support return of list object and accept projection json input
# sending the read method an empty document requests all documents be returned
df = pd.DataFrame.from_records(db.read({}))

# MongoDB v5+ is going to return the '_id' column and that is going to have an 
# invlaid object type of 'ObjectID' - which will cause the data_table to crash - so we remove
# it in the dataframe here. The df.drop command allows us to drop the column. If we do not set
# inplace=True - it will reeturn a new dataframe that does not contain the dropped column(s)
df.drop(columns=['_id'],inplace=True)

## Debug
# print(len(df.to_dict(orient='records')))
# print(df.columns)


#########################
# Dashboard Layout / View
#########################
app = Dash(__name__)

image_filename = 'Grazioso Salvare Logo.png' # replace with your own image
encoded_image = base64.b64encode(open(image_filename, 'rb').read())

app.layout = html.Div([
    html.Center(html.B(html.H1('Robert Rossman\'s CS-340 Dashboard'))),
    html.Hr(),
    # Place logo image and Radio buttons, input fields, and buttons side-by-side
    html.Div(style={'display' : 'flex'},
             children=[
        html.A(html.Img(src='data:image/png;base64,{}'.format(encoded_image.decode()),width="343",height="342"), href='https://data.austintexas.gov/Health-and-Community-Services/Austin-Animal-Center-Outcomes/gsvs-ypi7/data_preview'),
        html.Div([
            html.Div(style={'width': '300px'}, id='filter-type-label', children='Filter'),
            dcc.RadioItems(
                id= 'filter-type',
                options= [
                    {'label':'Water Rescue', 'value': 'water_rescue'},
                    {'label':'Mountain or Wilderness Rescue', 'value': 'mountain_rescue'},
                    {'label':'Disaster Rescue or Individual Tracking', 'value': 'disaster_rescue'},
                    {'label':'Reset', 'value': 'reset'}]
            ),
            # Display a column of buttons used to Create, Edit, and Delete entries
            html.Div(style={'display': 'grid', "width": "200px",'margin-top':'20px'},
                children=[
                html.Div(style={'margin-bottom': '5px'},id="admin-functions-label", children="Admin Functions"),
                dcc.Button('Create New Entry', id='create-entry-button', n_clicks=0),
                dcc.Button('Update Selected Entry', id='update-entry-button', n_clicks=0),
                dcc.Button('Delete Selected Entry', id='delete-entry-button', n_clicks=0),
            ]),
            html.Div(style={'margin-top': '20px'}, id= "admin-authorization-input-label", children="Admin Authorization"),
            dcc.Input(style={'margin-top': '5px', 'margin-bottom':'10px', "width": "200px"}, id= "admin-authorization-input", type='password'),
            html.Div(style={"color":"red"}, id= "admin-authorization-log", children=""),
        ]),
        # Create a column of text fields with labels placed side-by-side
        # These will be filled with the values of the entry that is currently selected
        html.Div([
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='rec-num-display-label', children='rec_num'),
                        html.Div(style={'width': '250px'}, id='rec-num-display'),
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='age-upon-outcome-input-label', children='age_upon_outcome'),
                        dcc.Input(id='age-upon-outcome-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='animal-id-input-label', children='animal_id'),
                        dcc.Input(id='animal-id-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='animal-type-input-label', children='animal_type'),
                        dcc.Input(id='animal-type-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='breed-input-label', children='breed'),
                        dcc.Input(id='breed-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='color-input-label', children='color'),
                        dcc.Input(id='color-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='date-of-birth-input-label', children='date_of_birth'),
                        dcc.Input(id='date-of-birth-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='datetime-input-label', children='datetime'),
                        dcc.Input(id='datetime-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='monthyear-input-label', children='monthyear'),
                        dcc.Input(id='monthyear-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='name-input-label', children='name'),
                        dcc.Input(id='name-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='outcome-subtype-input-label', children='outcome_subtype'),
                        dcc.Input(id='outcome-subtype-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='outcome-type-input-label', children='outcome_type'),
                        dcc.Input(id='outcome-type-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='sex-upon-outcome-input-label', children='sex_upon_outcome'),
                        dcc.Input(id='sex-upon-outcome-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='location-lat-input-label', children='location_lat'),
                        dcc.Input(id='location-lat-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='location-long-input-label', children='location_long'),
                        dcc.Input(id='location-long-input', type='text', style={"width": "250px"})
                ]),
            html.Div(style={'display' : 'flex'},
                    children=[
                        html.Div(style={'width': '200px'}, id='age-upon-outcome-in-weeks-input-label', children='age_upon_outcome_in_weeks'),
                        dcc.Input(id='age-upon-outcome-in-weeks-input', type='text', style={"width": "250px"})
                ])
        ])
    ]),
    html.Hr(),
    dash_table.DataTable(id='datatable-id',
                         columns=[{"name": i, "id": i, "deletable": False, "selectable": True} for i in df.columns],
                         data=df.to_dict('records'),
                         # Descriptions of properties taken from https://dash.plotly.com/datatable/reference
                         row_selectable = 'single',    # Allows users to select a single row with a button
                         column_selectable = 'single', # Allows users to select a single column with a button
                         selected_rows = [0],          # Selects an initial row
                         selected_columns = [],        # Starts with no column selected
                         filter_action = 'native',     # Makes filtering logic be "handled by the table" 
                         sort_action = 'native',       # Displays sorting UI and has it be "handled by the table"
                         sort_mode = 'multi',          # Allows sorting to be "performed across multiple columns"
                         page_size = 10,               # Sets the "number of rows that will be displayed" on one page
                        ),
    html.Br(),
    html.Hr(),
# This sets up the dashboard so that your chart and your geolocation chart are side-by-side
    html.Div(
         style={'display' : 'flex'},
             children=[
        html.Div(
            id='graph-id',
            className='col s12 m6',

            ),
        html.Div(
            id='map-id',
            className='col s12 m6',
            )
        ])
])

#############################################
# Interaction Between Components / Controller
#############################################

# Used to authorize admin actions
def authorize(passwordInput): 
    return password == passwordInput

    
@app.callback(Output('datatable-id', 'data'),
              [Input('filter-type', 'value')])
def update_dashboard(filter_type):
    # Start case
    # Also used if filter_type is 'reset'
    filter = {}
    
    # Set filter if filter_type is valid
    if filter_type == 'water_rescue':
        filter = {"animal_type": "Dog", "breed": {"$in": ["Labrador Retriever", "Newfoundland"]}}
    elif filter_type == 'mountain_rescue':
        filter = {"animal_type": "Dog", "breed": {"$in": ["German Shepherd", "Alaskan Malamute", "Old English Sheepdog", "Siberian Husky", "Rottweiler"]}}
    elif filter_type == 'disaster_rescue':
        filter = {"animal_type": "Dog", "breed": {"$in": ["Doberman Pinscher", "German Shepherd", "Golden Retriever", "Bloodhound", "Rottweiler"]}}
        
    df = pd.DataFrame.from_records(db.read(filter))
    # Cleanup Mongo _id field
    df.drop(columns=['_id'], inplace=True)
    return df.to_dict('records')

# Display the breeds of animal based on quantity represented in
# the data table
@app.callback(
    Output('graph-id', "children"),
    [Input('datatable-id', "derived_virtual_data")])
def update_graphs(viewData):
    dff = pd.DataFrame.from_dict(viewData)
    return [
       dcc.Graph(            
           figure = px.pie(dff, names='breed', title='Preferred Animal Breeds')
       )    
    ]
    
#This callback will highlight a cell on the data table when the user selects it
@app.callback(
    Output('datatable-id', 'style_data_conditional'),
    [Input('datatable-id', 'selected_columns')]
)
def update_styles(selected_columns):
    return [{
        'if': { 'column_id': i },
        'background_color': '#D2F3FF'
    } for i in selected_columns]


# This callback will update the geo-location chart for the selected data entry
# derived_virtual_data will be the set of data available from the datatable in the form of 
# a dictionary.
# derived_virtual_selected_rows will be the selected row(s) in the table in the form of
# a list. For this application, we are only permitting single row selection so there is only
# one value in the list.
# The iloc method allows for a row, column notation to pull data from the datatable
@app.callback(
    Output('map-id', "children"),
    [Input('datatable-id', "derived_virtual_data"),
     Input('datatable-id', "derived_virtual_selected_rows")])
def update_map(viewData, index):  
    if viewData is None:
        return
    elif index is None:
        return
    
    dff = pd.DataFrame.from_dict(viewData)
    # Because we only allow single row selection, the list can be converted to a row index here
    if index is None:
        row = 0
    else: 
        row = index[0]
        
    # Austin TX is at [30.75,-97.48]
    return [
        dl.Map(style={'width': '1000px', 'height': '500px'}, center=[30.75,-97.48], zoom=10, children=[
            dl.TileLayer(id="base-layer-id"),
            # Marker with tool tip and popup
            # Column 13 and 14 define the grid-coordinates for the map
            # Column 4 defines the breed for the animal
            # Column 9 defines the name of the animal
            dl.Marker(position=[dff.iloc[row,13],dff.iloc[row,14]], children=[
                dl.Tooltip(dff.iloc[row,4]),
                dl.Popup([
                    html.H1("Animal Name"),
                    html.P(dff.iloc[row,9])
                ])
            ])
        ])
    ]


# This callback will update the input fields for the selected data entry
# derived_virtual_data will be the set of data available from the datatable in the form of 
# a dictionary.
# derived_virtual_selected_rows will be the selected row(s) in the table in the form of
# a list. For this application, we are only permitting single row selection so there is only
# one value in the list.
# The iloc method allows for a row, column notation to pull data from the datatable
@app.callback(
    [Output('rec-num-display', "children"),
     Output('age-upon-outcome-input', "value"),
     Output('animal-id-input', "value"),
     Output('animal-type-input', "value"),
     Output('breed-input', "value"),
     Output('color-input', "value"),
     Output('date-of-birth-input', "value"),
     Output('datetime-input', "value"),
     Output('monthyear-input', "value"),
     Output('name-input', "value"),
     Output('outcome-subtype-input', "value"),
     Output('outcome-type-input', "value"),
     Output('sex-upon-outcome-input', "value"),
     Output('location-lat-input', "value"),
     Output('location-long-input', "value"),
     Output('age-upon-outcome-in-weeks-input', "value")],
    [Input('datatable-id', "derived_virtual_data"),
     Input('datatable-id', "derived_virtual_selected_rows")])
def update_text_input_fields(viewData, index):  
    if viewData is None:
        return
    elif index is None:
        return
    
    dff = pd.DataFrame.from_dict(viewData)
    # Because we only allow single row selection, the list can be converted to a row index here
    if index is None:
        row = 0
    else: 
        row = index[0]
        
    # Austin TX is at [30.75,-97.48]
    return [
        dff.iloc[row,0],  # rec_num
        dff.iloc[row,1],  # age_upon_outcome
        dff.iloc[row,2],  # animal_id
        dff.iloc[row,3],  # animal_type 
        dff.iloc[row,4],  # breed 
        dff.iloc[row,5],  # color
        dff.iloc[row,6],  # date_of_birth
        dff.iloc[row,7],  # datetime
        dff.iloc[row,8],  # monthyear
        dff.iloc[row,9],  # name
        dff.iloc[row,10], # outcome_subtype
        dff.iloc[row,11], # outcome_type
        dff.iloc[row,12], # sex_upon_outcome
        dff.iloc[row,13], # location_lat
        dff.iloc[row,14], # location_long
        dff.iloc[row,15]  # age_upon_outcome_in_weeks
        ]

# def get_data_from_values(age_upon_outcome, animal_id, animal_type, breed, color, 
#                             date_of_birth, datetime, monthyear, name, outcome_subtype, outcome_type, 
#                             sex_upon_outcome, location_lat, location_long, age_upon_outcome_in_weeks):
#     return {{"age_upon_outcome": age_upon_outcome, "animal_id": animal_id, "animal_type": animal_type, 
#             "breed": breed, "color": color, "date_of_birth": date_of_birth, "datetime": datetime, "monthyear": monthyear,
#             "name": name, "outcome_subtype": outcome_subtype, "outcome_type": outcome_type, "sex_upon_outcome": sex_upon_outcome,
#             "location_lat": location_lat, "location_long": location_long, "age_upon_outcome_in_weeks": age_upon_outcome_in_weeks}}

# This callback will create a new entry using the values in the input fields
# A new rec_num will automatically be assigned, as rec_nums must be unique
# Displays will be updated by forcing an update_dashboard callback
@app.callback(
    Output('filter-type', 'value', allow_duplicate=True), # Use allow_duplicate for other callbacks with same outputs
    Output('admin-authorization-input', 'value', allow_duplicate=True),
    Output('admin-authorization-log', 'children', allow_duplicate=True),
    Output('admin-authorization-log', 'style', allow_duplicate=True),
    Input('create-entry-button', 'n_clicks'),
    [State('admin-authorization-input', 'value'),
     State('age-upon-outcome-input', 'value'),
     State('animal-id-input', 'value'),
     State('animal-type-input', 'value'),
     State('breed-input', 'value'),
     State('color-input', 'value'),
     State('date-of-birth-input', 'value'),
     State('datetime-input', 'value'),
     State('monthyear-input', 'value'),
     State('name-input', 'value'),
     State('outcome-subtype-input', 'value'),
     State('outcome-type-input', 'value'),
     State('sex-upon-outcome-input', 'value'),
     State('location-lat-input', 'value'),
     State('location-long-input', 'value'),
     State('age-upon-outcome-in-weeks-input', 'value'),
     State('filter-type', 'value')],
    prevent_initial_call=True
)
def create_entry(n_clicks, admin_authorization_input,
                 age_upon_outcome, animal_id, animal_type, breed, color, 
                 date_of_birth, datetime, monthyear, name, outcome_subtype, outcome_type, 
                 sex_upon_outcome, location_lat, location_long, age_upon_outcome_in_weeks, 
                 filter_type):
    output_log = ""
    # Set output_log color to red by default
    output_log_style = {"color": "red"}
    if (authorize(admin_authorization_input)):
        # Create a new entry using values from inputs
        if db.create({ "age_upon_outcome": age_upon_outcome, "animal_id": animal_id, "animal_type": animal_type, "breed": breed, "color": color, 
                    "date_of_birth": date_of_birth, "datetime": datetime, "monthyear": monthyear, "name": name, "outcome_subtype": outcome_subtype,
                    "outcome_type": outcome_type, "sex_upon_outcome": sex_upon_outcome, "location_lat": location_lat, "location_long": location_long,
                    "age_upon_outcome_in_weeks": age_upon_outcome_in_weeks}):
                    output_log = "Created entry with rec_num " + str(db.find_next_rec_num() - 1)
                    # Set output_log color to green to indicate success
                    output_log_style["color"] = "green"
        else:
            output_log = "Failed to create entry"
    else:
        output_log = "Entry creation not authorized"
    # 1. Update the displays by refreshing the filter-type
    # radio buttons to force an update_dashboard callback
    # 2. Clear the password input field
    # 3. Update the output_log to indicate success/failure
    return filter_type, "", output_log, output_log_style

# This callback will update the currently selected entry
# Values in the input fields will be used to update the entry
# The rec_num field will remain unchanged as it must be unique
# Displays will be updated by forcing an update_dashboard callback
@app.callback(
    Output('filter-type', 'value', allow_duplicate=True),
    Output('admin-authorization-input', 'value', allow_duplicate=True),
    Output('admin-authorization-log', 'children', allow_duplicate=True),
    Output('admin-authorization-log', 'style', allow_duplicate=True),
    Input('update-entry-button', 'n_clicks'),
    [State('admin-authorization-input', 'value'),
     State('rec-num-display', 'children'),
     State('age-upon-outcome-input', 'value'),
     State('animal-id-input', 'value'),
     State('animal-type-input', 'value'),
     State('breed-input', 'value'),
     State('color-input', 'value'),
     State('date-of-birth-input', 'value'),
     State('datetime-input', 'value'),
     State('monthyear-input', 'value'),
     State('name-input', 'value'),
     State('outcome-subtype-input', 'value'),
     State('outcome-type-input', 'value'),
     State('sex-upon-outcome-input', 'value'),
     State('location-lat-input', 'value'),
     State('location-long-input', 'value'),
     State('age-upon-outcome-in-weeks-input', 'value'),
     State('filter-type', 'value')],
    prevent_initial_call=True
)
def update_entry(n_clicks, admin_authorization_input, rec_num, 
                 age_upon_outcome, animal_id, animal_type, breed, color, 
                 date_of_birth, datetime, monthyear, name, outcome_subtype, outcome_type, 
                 sex_upon_outcome, location_lat, location_long, age_upon_outcome_in_weeks, 
                 filter_type):
    output_log = ""
    # Set output_log color to red by default
    output_log_style = {"color": "red"}
    if (authorize(admin_authorization_input)):
        # Create a filter to get the matching entry in the database
        filter = {"rec_num": rec_num}
        # Update any records that match the filter
        if (db.update(filter, {"$set":{"age_upon_outcome": age_upon_outcome, "animal_id": animal_id, "animal_type": animal_type, "breed": breed, "color": color, 
                                "date_of_birth": date_of_birth, "datetime": datetime, "monthyear": monthyear, "name": name, "outcome_subtype": outcome_subtype,
                                "outcome_type": outcome_type, "sex_upon_outcome": sex_upon_outcome, "location_lat": location_lat, "location_long": location_long,
                                "age_upon_outcome_in_weeks": age_upon_outcome_in_weeks}}) > 0):
                                output_log = "Updated entry with rec_num" + str(rec_num)
                                # Set output_log color to green to indicate success
                                output_log_style["color"] = "green"
        else:
            output_log = "No entry with rec_num " + str(rec_num) + " found to update"
    else:
        output_log = "Entry update not authorized"
    # 1. Update the displays by refreshing the filter-type
    # radio buttons to force an update_dashboard callback
    # 2. Clear the password input field
    # 3. Update the output_log to indicate success/failure
    return filter_type, "", output_log
    

# This callback will delete the currently selected entry
# Displays will be updated by forcing an update_dashboard callback
@app.callback(
    Output('filter-type', 'value'),
    Output('admin-authorization-input', 'value'),
    Output('admin-authorization-log', 'children'),
    Output('admin-authorization-log', 'style'),
    Input('delete-entry-button', 'n_clicks'),
    [State('admin-authorization-input', 'value'),
     State('rec-num-display', 'children'),
     State('filter-type', 'value')],
    prevent_initial_call=True
)
def delete_entry(n_clicks, admin_authorization_input, rec_num, filter_type):
    output_log = ""
    # Set output_log color to red by default
    output_log_style = {"color": "red"}
    if (authorize(admin_authorization_input)):
        # Create a filter to get the matching entry in the database
        filter = {"rec_num": rec_num}
        # Delete any records that match the filter
        if(db.delete(filter) > 0):
            output_log = "Deleted entry with rec_num " + str(rec_num)
            # Set output_log color to green to indicate success
            output_log_style["color"] = "green"
        else:
            output_log = "No entry with rec_num " + str(rec_num) + " found to delete"
    else:
        output_log = "Entry deletion not authorized"
    # 1. Update the displays by refreshing the filter-type
    # radio buttons to force an update_dashboard callback
    # 2. Clear the password input field
    # 3. Update the output_log to indicate success/failure
    return filter_type, "", output_log

# Run app and display result in jupyterlab mode, note, if you have previously run a prior app, the default port of 8050 may not be available, if so, try setting an alternate port.
app.run() 
```
